```python
%%tikz -p=\include{prefix.tex} -sc=2 -S laplace_integral_calc.svg
\begin{tikzpicture}[auto, node distance=1.5cm,>=latex]
    \node[yshift=3cm]  {
        \begin{minipage}{\textwidth}
            Gitt at vi har et lineært og tids-invariant system med impulsrespons $h(t)$, kan vi utlede hva utgangen blir når vi påtrykker egenfunksjonen $u(t) = e^{s\cdot t}$:
            \begin{enumerate}
            \item Uttrykk systemet slik at utgangen $y(t)$ kan utrrykkes som en funksjon av påtrykket $u(t)$:
                \begin{equation*}y(t) = F(u(t))\end{equation*}
            \item Med litt triksing kan vi omskrive $u(t)$ til et integraluttrykk med forskyving av deltapulsen $\delta (t)$:
                \begin{align*}
                    u(t) &= \int_{-\infty}^{\infty} u(\tau) \cdot \delta(t-\tau) d\tau \\
                    &\Downarrow \\
                    y(t) &= F\left( \int_{-\infty}^{\infty} u(\tau) \cdot \delta(t-\tau) d\tau \right)
                \end{align*}
            \item Videre er inngår det av linearitetsegenskapene til systemet at vi kan sette integralet utenfor funksjonsuttrykket $F(\cdots)$, ettersom $F(a\cdot u_1(t) + b \cdot u_2(t)) = a \cdot F(u_1(t)) + b \cdot  F( u_2(t))$
                \begin{equation*} F\left( \int_{-\infty}^{\infty} u(\tau) \cdot \delta(t-\tau) d\tau \right) = \int_{-\infty}^{\infty} u(\tau) \cdot F\left( \delta(t-\tau) \right) d\tau \end{equation*}
            \item Definisjonen på systemets impulsrespons er $h(t) = F(\delta(t))$, og siden systemet er tids-invariant, betyr det også at $F(\delta(t-\tau)) = h(t-\tau)$.
                \begin{equation*} y(t) = \int_{-\infty}^{\infty} u(\tau) \cdot h(t-\tau) d\tau \end{equation*}
            \item Uttrykket ovenfor er det vi kaller et konvolusjonsintegral, og ettersom konvolusjon er en kommutativ operasjon kan vi bytte plass på uttrykkene inne i integralet.
                \begin{equation*} y(t) = \int_{-\infty}^{\infty} h(\tau) \cdot u(t-\tau) d\tau \end{equation*}
            \item Nå kan vi gjøre en viss antagelse om at universet systemet eksisterer i er kausalt. Det vil si at dersom systemet påtrykkes en deltapuls, er det på ingen måte mulig at systemet produserer en respons \begin{textit}før\end{textit} impulsen treffer systemet. Vi kan derfor konkludere at $h(t) = 0$ for alle verdier $t < 0$.
                \begin{align*}
                y(t) &= \int_{-\infty}^{0} h(\tau) \cdot u(t-\tau) d\tau + \int_{0}^{\infty} h(\tau) \cdot u(t-\tau) d\tau\\
                &= \int_{0}^{\infty} h(\tau) \cdot u(t-\tau) d\tau
                \end{align*}
            \item Til slutt kan vi utføre substitusjonene $u(t) = e^{s\cdot t}$ og $y(t) = \lambda(s) \cdot e^{s\cdot t}$:
                \begin{align*}
                y(t) &= \int_{0}^{\infty} h(\tau) \cdot u(t-\tau) d\tau \\
                \lambda(s) \cdot e^{s\cdot t} &= \int_{0}^{\infty} h(\tau) \cdot e^{s\cdot (t-\tau)} d\tau \\
                \lambda(s) \cdot e^{s\cdot t} &= \int_{0}^{\infty} h(\tau) \cdot e^{s\cdot t } \cdot e^{-s \cdot \tau} d\tau \\
                \lambda(s) &= \int_{0}^{\infty} h(\tau)\cdot e^{-s \cdot \tau} d\tau
                \end{align*}
            \end{enumerate}
        Det vi sitter igjen med er et uttrykk for hva egenverdien $\lambda$ til systemet er gitt en egenfunksjon $e^{s\cdot t} $. Dette uttrykket er uttrykket for laplacetransformasjon.
        \end{minipage}
    };
\end{tikzpicture}
```
