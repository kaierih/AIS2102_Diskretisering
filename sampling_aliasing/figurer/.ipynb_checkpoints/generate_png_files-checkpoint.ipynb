```python
import numpy as np
import matplotlib.pyplot as plt

T = 1
t = np.linspace(0, T, 101)

ut = np.sin(np.pi*2*t)

fs = 8
t_k = np.arange(0, T + 1/fs, 1/fs)
uk = np.sin(2*np.pi*t_k)

plt.figure(figsize=(2,2))
plt.stairs(uk[:-1], t_k)
plt.plot(t, ut, ":", linewidth="0.5", color="tab:red")
plt.xticks([])
plt.yticks([])
```
