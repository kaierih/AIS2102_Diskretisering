# AIS2102 - Diskretiseringsmodul

Denne samlingen med Jupyter Notatbøker utgjør undervisningsnotatene benyttet i forelesningene om Diskretisering i emnet AIS2102 - Dynamiske Systemer. Dokumentene benytter i Python-kode for å tilføre interaktive illustrasjoner og kodeeksempler for å styrke sammenhengen mellom matematisk teori og anvendelse med programmering. I en del tilfeller vil også tilsvarende MATLAB-kode for kodeeksempler oppgis, men dette vil ikke være som kjørbar kode. 

Forelesningsnotatene er inndelt i en kapittelstruktur etter tema, og hver jupyter notebook fil er tiltenkt å korrespondere med én $45$-minutters undervisningstime. Dokumentene vil oppdateres etter hvert som forelesningen utføres.
